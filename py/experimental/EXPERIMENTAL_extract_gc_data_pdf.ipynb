```python
import os
import tabula

def extract_tables_from_pdf(file):
    # Attempt to extract any tables from the PDF file
    tables = tabula.read_pdf(file, pages='all', multiple_tables=True)

    # Get the directory of the PDF file
    directory = os.path.dirname(file)

    # If there are no tables, the list will be empty
    if not tables:
        print('No tables found in the PDF file.')
        return None
    else:
        # If tables were found, write each of them to a separate CSV file in the same directory as the PDF file
        for i, table in enumerate(tables):
            table.to_csv(os.path.join(directory, f'table_{i}.csv'), index=False)

        print(f'{len(tables)} tables have been written to CSV files in the directory: {directory}.')
```

```python
import os
import csv
import pdfplumber

def extract_text_from_pdf(file):
    # Open the PDF file
    with pdfplumber.open(file) as pdf:
        # Get the directory of the PDF file
        directory = os.path.dirname(file)

        # Open a new CSV file in the same directory as the PDF file
        with open(os.path.join(directory, 'output.csv'), 'w', newline='') as csvfile:
            writer = csv.writer(csvfile)

            # Go through each page in the PDF
            for i, page in enumerate(pdf.pages):
                # Extract text from the page
                text = page.extract_text()

                # Split the text into lines and write each line to the CSV file
                for line in text.split('\n'):
                    writer.writerow([line])

    print(f'Text has been written to output.csv in the directory: {directory}.')
extract_text_from_pdf(r'H:\data\drm\ugent\sasol-1.8-c450\chromatograms\05.0s-01-blank.rslt\05.0s-01-blank-Rep1_1_MCR.pdf')
```
