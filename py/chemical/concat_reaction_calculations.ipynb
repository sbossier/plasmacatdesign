```python
import numpy as np
import pandas as pd
import os
```

```python
# Specify the parent directory
parent_dir = r'N:\FWET\FDCH\AdsCatal\General\personal_work_folders\plasmacatdesign\co2-splitting\uhasselt'

# List for collecting all data
data_list = []

# Walk through each directory and its subdirectories
for dirpath, dirnames, filenames in os.walk(parent_dir):
    # For each file
    if dirnames == "ignore":
        continue
    else:
        for filename in filenames:
            # If the file ends in '-calc-short.csv'
            if filename.endswith('-calc-short.csv'):
                # Construct full file path
                file_path = os.path.join(dirpath, filename)
                # Read the CSV file and append it to the list
                data = pd.read_csv(file_path, keep_default_na=True)
                
                # Extract the measurement number from the filename
                # and format it as a number
                measurement_number = filename.split('-')[1]
                
                # Add the measurement number as a new column
                data['measurement_number'] = measurement_number
                
                # Append the data to the list
                data_list.append(data)

# Concatenate all data frames
combined_data = pd.concat(data_list, ignore_index=True)

# Fill NA values with 'NA'
combined_data.fillna(np.nan, inplace=True)

# Save to new csv file in the parent directory
combined_data.to_parquet(
    os.path.join(parent_dir, 'uhasselt_co2_spl_reaction_data_combined.parquet'),
    index=False,
    compression='gzip'
)

combined_data.to_csv(
	os.path.join(parent_dir, 'uhasselt_co2_spl_reaction_data_combined.csv'),
	index=False
)

print("All short files have been successfully combined!")
```
