```python
import pandas as pd
import glob
import re

def process_csv_files(parent_folder):
    # Construct the pattern for glob to match files
    pattern = parent_folder + "/*[0-9][0-9].[0-9]s-[0-9][0-9].csv"
    
    # Find all files that match the pattern
    csv_files = glob.glob(pattern)
    
    # Compile a regex pattern to strictly match the "dd.ds-dd.csv" pattern
    strict_pattern = re.compile(r'\d{2}\.\d{1}s-\d{2}\.csv$')
    
    for file_path in csv_files:
        # Check if the file strictly matches the "dd.ds-dd.csv" pattern
        if strict_pattern.search(file_path):
            # Read the CSV file into a DataFrame
            df = pd.read_csv(file_path)
            
            # Create the 'gas_temp_in_dgc' column if it does not exists
            if 'gas_temp_in_dgc' not in df.columns:
                df['gas_temp_in_dgc'] = None
                
            # Replace 'gas_temp_in_dgc' values based on 'state' conditions
            df.loc[df['state'] == 'blank', 'gas_temp_in_dgc'] = 20
            df.loc[df['state'] == 'plasma', 'gas_temp_in_dgc'] = 70
            
            # Replace 'gas_temp_out_dgc' values based on 'state' conditions
            df.loc[df['state'] == 'blank', 'gas_temp_out_dgc'] = 20
            df.loc[df['state'] == 'plasma', 'gas_temp_out_dgc'] = 70
            
            # Save the modified DataFrame back to the CSV, overwriting the original file
            df.to_csv(file_path, index=False)
            print(f"Processed and updated file: {file_path}")
```

```python
# Example usage
parent_folder = r"N:\FWET\FDCH\AdsCatal\General\personal_work_folders\plasmacatdesign\co2-splitting\uhasselt\SiO2+TMAH-220-03H\pwr-const"
process_csv_files(parent_folder)
```
